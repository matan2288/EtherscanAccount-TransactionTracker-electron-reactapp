```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

הבעיה העסקית: זיהוי גורמים המשפיעים על ההצלחה הכלכלית בקרב 1000 הסרטים המובילים ב IMDB.

הדאטה סט מכיל את הסרטים בעלי הדירוג הגבוה ביותר. המטרה היא להבין האם קיים קשר בין מאפייני הסרט לבין ההכנסות שלו.
לשם כך נבצע ניקוי נתונים הכולל המרת נתוני ההכנסות וזמן הסרט למספרים ונטפל בערכים חסרים כדי שנוכל לעשות ניתוח סטטיסטי

```python
df = pd.read_csv('imdb_top_1000.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
# check for duplicate rows
dup_count = df.duplicated().sum()
print(f"מספר השורות הכפולות בדאטה הוא: {dup_count}")
```

```python
# drop duplicate rows
# inplace = true is for the change is saved to the current dataframe
df.drop_duplicates(inplace=True)
```

```python
# rename columns name
df.rename(columns={'Series_Title' : 'Movie_Name'}, inplace = True)
df.head(2)
```

```python
# remove the string 'min' and convert to integer
df['Runtime'] = df['Runtime'].astype(str).str.replace(' min', '').astype(int)
# remove commas from Gross column (e.g. '1,000' to '1000')
df['Gross'] = df['Gross'].str.replace(',', '')
#convert to numeric and 'coerce' turns invalid values into NaN instead of crashing the code
# the converted to numeric is to allow calculation of median
df['Gross']= pd.to_numeric(df['Gross'], errors = 'coerce')
df.info()
```

```python
#calculate the median of 'Gross' revenue
Gross_median = df['Gross'].median()
# fill missing values in 'Gross' with the calculated median
df['Gross'] = df['Gross'].fillna(Gross_median)
# check if ther are any missing values in 'Gross'
print(f"The missing values in Gross is: {df['Gross'].isnull().sum()}")
```

```python
df.info()
```

```python
# cleaning the remaining columns
# Meta_Score: fill missing values wuth the median
Meta_median = df['Meta_score'].median()
df['Meta_score'] = df['Meta_score'].fillna(Meta_median)
#Certificate : fill missing values with 'Unknown'
df['Certificate'] = df['Certificate'].fillna('Unknown')

df.info()
```

```python
# set figure size
plt.figure(figsize=(10,5))
# create a histogram fo IMDB Rating
sns.histplot(df['IMDB_Rating'], bins =15, kde=True)
# add titles and labels
plt.title('Distribution of IMDB Ratings')
plt.xlabel('Rating')
plt.ylabel('Count')

plt.show()
```

```python
# split the string of 'Genres' by comma and take only the first Genre
df['main_Genre'] = df['Genre'].astype(str).str.split(',').str[0]
plt.figure(figsize=(12,6))
#show the count of movies for the top 10 Genres
sns.countplot(
    data= df,
    x= 'main_Genre',
    order=df['main_Genre'].value_counts().index[:10])
plt.title('Top 10 most common Genres')
plt.xlabel('Genre')
plt.ylabel('Number of movies')
plt.xticks(rotation=45)

plt.show()
```

```python
# **Extra- with help from Gemini**
# Multi-variate Analysis (Rating vs. Revenue vs. Genre)

# Creat a new column for Revenue in millions
df['Gross_millions'] = df['Gross']/ 1000000
# set figure size
plt.figure(figsize=(12,8))
# Scatterplot using 3 variables:
# x = Rating
# y = Gross
# Hue = Genre

sns.scatterplot(
    data= df,
    x= 'IMDB_Rating',
    y= 'Gross_millions',
    hue= 'main_Genre',
    s =100, #Size of dots
    alpha= 0.7 # Transparency (help when dots overlap)
    )
plt.title('Business Insight: Revenue vs. Rating by Genre')
plt.xlabel('IMDB Rating')
plt.ylabel('Gross Revenue (in millions)')
plt.legend(bbox_to_anchor=(1.05,1), loc= "upper left", title='Genre')
plt.grid(True, linestyle ='--', alpha=0.3)

plt.show()
```
